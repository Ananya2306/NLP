*EXPERIMENT 1.1 -  TEXT PREPROCESSING PIPELINE*

```python
import re
import nltk
from nltk import pos_tag
from nltk.corpus import stopwords, wordnet
from nltk.stem import WordNetLemmatizer, PorterStemmer
from nltk.tokenize import word_tokenize, sent_tokenize

for pkg in ["punkt", "punkt_tab", "stopwords", "wordnet", "averaged_perceptron_tagger", "averaged_perceptron_tagger_eng"]:
  nltk.download(pkg)

text = ("Dr. Rao's team is building chatbots, translators and search engines! "
        "The systems were running faster than expected. "
        "Visit https://example.com or mail nlp@lab.edu for details.")
# Step 1: sentence segmentation (before lower-casing so that 'Dr.' is handled)
sentences = sent_tokenize(text)
print("Sentences:", len(sentences))
for s in sentences:
       print("  -", s)

 # Step 2: normalisation - remove URLs / e-mails, lower-case, strip punctuation
def normalise(s):
      s = re.sub(r"https?://\S+|\S+@\S+", " ", s)
      return s.lower()

  # Step 3: word tokenisation and stop-word removal
stop = set(stopwords.words("english"))
tokens = [t for s in sentences for t in word_tokenize(normalise(s))]
words = [t for t in tokens if t.isalpha()]
content = [t for t in words if t not in stop]
print("\nTokens  :", tokens)
print("Content :", content)

# Step 4: stemming vs lemmatisation (POS-aware)
ps, wnl = PorterStemmer(), WordNetLemmatizer()
def to_wn(tag):
      return {"J": wordnet.ADJ, "V": wordnet.VERB,
               "N": wordnet.NOUN, "R": wordnet.ADV}.get(tag[0], wordnet.NOUN)

print(f"\n{'word':12}{'stem':12}{'lemma':12}")
for w, tag in pos_tag(content):
         print(f"{w:12}{ps.stem(w):12}{wnl.lemmatize(w, to_wn(tag)):12}")

# Step 5: vocabulary reduction
print("\nVocabulary size  raw:", len(set(words)),
        "| after stop-word removal:", len(set(content)),
        "| after stemming:", len({ps.stem(w) for w in content}))
```
